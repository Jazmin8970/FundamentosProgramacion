```python
tabla = int (input("¿Qué tabla quieres? "))
lim1 = int (input("¿Cuál es el limite inicial? "))
lim2 = int (input("¿Cuál es el limite final? "))
inte = int (input("¿Cuál es el intervalo? "))

# agregamos la instruccion while para que valide que el valor iniciar sea mayor que el final, de lo contrario me pedirá indefinidamente el valor final hasta que sea mayor
while lim1 >= lim2:
  print("El limite final debe ser mayor que el inicio")
  lim2 = int(input ("¿Cuál es el limite final? "))

for i in range(lim1, lim2, inte):
  print(tabla, "*", i, "=", tabla*i)
```
