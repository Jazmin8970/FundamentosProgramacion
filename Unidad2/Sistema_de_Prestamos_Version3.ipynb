```python
import datetime
#=========================================
# SISTEMA DE PRÉSTAMOS
# Versión 3 - Uso de funciones y listas de listas
#=========================================

alumnos = [
    ["26040200","Alejandra García", "Ingeniería en Inteligencia Artificial"],
    ["26040201","Michelle Dominguez", "Ingeniería en Inteligencia Artificial"],
    ["26040202", "Gabriel Carrizosa", "Ingeniería Bioquímica"],
    ["26040203","Oscar Lozano", "Ingeniería en Sistemas Computacionales"],
    ["26040204", "Claudia López", "Ingeniería Biomédica"]
]


computadoras = [
                ["PC-LC1-01", "LC1", "Python, VS Code", "Disponible"],
                ["PC-LC1-02", "LC1", "Python, Java, VS Code", "Disponible"],
                ["PC-LC1-03", "LC1", "AutoCAD, MATLAB", "Reservada"],
                ["PC-LC2-01", "LC2", "Python, VS Code", "Disponible"],
                ["PC-LC2-02", "LC2", "Python, Java, VS Code", "Disponible"],
                ["PC-LC2-03", "LC2", "AutoCAD, MATLAB", "Disponible"],
                ["PC-LC3-01", "LC3", "Python, VS Code", "Disponible"],
                ["PC-LC3-02", "LC3", "Python, Java, VS Code", "Prestada"],
                ["PC-LC3-03", "LC3", "AutoCAD, MATLAB", "Reservada"],
                ["PC-LC3-04", "LC3", "Python, VS Code", "Mantenimiento"]
              ]


prestamos = []

necesidades = []

#1
def listar_computadoras():
   print("\n--- COMPUTADORAS ---")
   for i in range(len(computadoras)):
      print(i+1, " Aula: ", computadoras[i][1], "| ID: ", computadoras[i][0], "|Software: ", computadoras[i][2], "|Estado: ", computadoras[i][3])
      print("------------------------")


#2
def buscar_sw():
  programa = input("\nIngrese el software que desea buscar: ")
  print("\n--- RESULTADOS ---")
  encontrado = False
  for i in range(len(computadoras)):
      # la funcion (lower) convierte a minuscula el texto, de esta forma
      # nos aseguramos que el texto a buscar y la informacion de las listas
      # este en minuscula y no tengamos problema de que no encuentre un programa
      # porque este escrito de forma diferente usando mayusculas y minusculas
      if programa.lower() in computadoras[i][2].lower():
          print(i, " Aula:", computadoras[i][1],"ID:", computadoras[i][0],"Software:", computadoras[i][2],"Estado:", computadoras[i][3])
          encontrado = True

  if encontrado == False:
      print("No se encontraron computadoras con ese software.")



def solicitar_prestamo():
    print("\n--- SOLICITAR PRÉSTAMO ---")
    num_control = input("Ingrese el número de control: ")


    existe_alumno = False
    for alu in alumnos:
        if alu[0] == num_control:
            existe_alumno = True
            break

    if existe_alumno:
        programa = input("Que programa necesita: ")

        # Buscar si la computadora tiene el programa
        encontrado = False
        for pc in computadoras:
            if programa.lower() in pc[2].lower():
                encontrado = True
                if pc[3] == "Disponible":
                    pc[3] = "Prestada"
                    print("Se asignó la computadora", pc[0], "que esta en el aula", pc[1])
                    prestamos.append([num_control, programa, pc[1], pc[0]])
                    break
                else:
                    print("Si se encontró el software pero está", pc[3])

        # Si el programa NO existe
        if not encontrado:
            resp = input("El programa no existe. ¿Desea agregarlo a necesidades? (s/n): ")
            if resp.lower() == 's':
                fecha = datetime.datetime.now().strftime("%d/%m/%Y %H:%M:%S")
                necesidades.append([programa, fecha])
                print("Agregado a la lista de necesidades.")
    else:
        print("El alumno no existe")



#4
def devolver_computadora():
  print("\n--- DEVOLVER COMPUTADORA ---")
  id_pc = input("Dame el ID de la computadora: ")
  existe = False

  for i in range(len(computadoras)):
    if computadoras[i][0] == id_pc:
      existe = True
      if computadoras[i][3] == "Prestada" or computadoras[i][3] == "Reservada":
        computadoras[i][3] = "Disponible"
        print("La computadora", id_pc, "ya está disponible")
      else:
        print("La computadora", id_pc, "no está prestada/reservada (Estado actual:", computadoras[i][3], ")")
      break

  if not existe:
    print("La computadora", id_pc, "no existe")


#6
def agregar_computadora():
  print("\n--- AGREGAR COMPUTADORA ---")
  id_computadora = input("ID de la computadora: ")
  aula = input("Aula: ")
  programas = input("Software instalado: ")
  nueva_compu = [id_computadora, aula, programas, "Disponible"]
  computadoras.append(nueva_compu)
  print("\nComputadora agregada correctamente.")

#7
def historial_prestamos():
  print("\n--- HISTORIAL DE PRESTAMOS ---")
  for prestamo in prestamos:
    print("Alumno:", prestamo[0], "| Software:", prestamo[1], "| Aula:", prestamo[2], "| ID:", prestamo[3])

#8
def historial_necesidades():
    print("\n--- HISTORIAL DE NECESIDADES ---")
    i = 1
    for nec in necesidades:
        print(i, "| Programa:", nec[0], "| Fecha:", nec[1], "|")
        i += 1

#9
def estadisticas():
   # --- Conteo por Carrera ---
  conteo_carreras = []
  for i in range(len(prestamos)):
    num_control = prestamos[i][0]
    carrera = ""
    # Buscar la carrera en la lista de alumnos
    for j in range(len(alumnos)):
      if alumnos[j][0] == num_control:
        carrera = alumnos[j][2]

        # Contar la carrera
        encontrado = False
        for k in range(len(conteo_carreras)):
          if conteo_carreras[k][0] == carrera:
            conteo_carreras[k][1] += 1
            encontrado = True

        if not encontrado:
          conteo_carreras.append([carrera, 1])

  print("\n======= Estadísticas por carrera ==========")
  for k in range(len(conteo_carreras)):
    print(conteo_carreras[k][0], ":", conteo_carreras[k][1], "veces")


  # --- Conteo por Programa ---
  conteo_programas = []
  for i in range(len(prestamos)):
    programa = prestamos[i][1]

    encontrado = False
    for j in range(len(conteo_programas)):
      if conteo_programas[j][0] == programa:
        conteo_programas[j][1] += 1
        encontrado = True

    if not encontrado:
      conteo_programas.append([programa, 1])

  print("\n======= Estadísticas por programa ===========")
  for i in range(len(conteo_programas)):
    print(conteo_programas[i][0], ":", conteo_programas[i][1], "veces")




# MENÚ PRINCIPAL

opcion = 0
while opcion <= 9:
  print("\n==========================")
  print("    SISTEMA DE PRESTAMOS    ")
  print("\n==========================")
  print("1. Consultar computadoras")
  print("2. Buscar por software")
  print("3. Solicitar prestamo")
  print("4. Devolver computadora")
  print("5. Crear reservación")
  print("6. Agregar computadora")
  print("7. Historial de Prestamos")
  print("8. Historial de Necesidades")
  print("9. Estadisticas")
  print("10. Salir")
  print("\n==========================")
  try:
    opcion = int(input("Seleccione una opción: "))
  except ValueError:
    print("Por favor ingrese un número válido.")
    continue

  # 1. Consultar computadora
  if opcion == 1:
     listar_computadoras()

  # 2. Buscar por software
  elif opcion == 2:
      buscar_sw()

  # 3. Solicitar préstamo
  elif opcion == 3:
    solicitar_prestamo()

  # 4. Devolver computadora
  elif opcion == 4:
      devolver_computadora()

  # 6. Agregar computadora
  elif opcion == 6:
      agregar_computadora()

  # 7. Historial de prestamos
  elif opcion == 7:
    historial_prestamos()

  # 8. Historial de necesidades
  elif opcion == 8:
    historial_necesidades()

  # 9. Estadisticas
  elif opcion == 9:
    estadisticas()
```
