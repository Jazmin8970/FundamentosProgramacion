```python
import datetime
#=========================================
# SISTEMA DE PRÉSTAMOS
# Versión 4 - Uso de Listas y Diccionarios
#=========================================

alumnos = [
    { "num_control": "26040200", "nombre": "Alejandra García",   "carrera":    "Ingeniería en Inteligencia Artificial"   },
    { "num_control": "26040201", "nombre": "Michelle Dominguez", "carrera":    "Ingeniería en Inteligencia Artificial"   },
    { "num_control": "26040202", "nombre":  "Gabriel Carrizosa", "carrera":    "Ingeniería Bioquímica"                   },
    { "num_control": "26040203", "nombre": "Oscar Lozano",       "carrera":    "Ingeniería en Sistemas Computacionales"  },
    { "num_control": "26040204", "nombre":  "Claudia López",     "carrera":    "Ingeniería Biomédica"                    }
]


computadoras = [
                { "id": "PC-LC1-01", "aula": "LC1",  "programas": "Python, VS Code",       "estado":  "Disponible"    },
                { "id": "PC-LC1-02", "aula": "LC1",  "programas": "Python, Java, VS Code", "estado":  "Prestada"      },
                { "id": "PC-LC1-03", "aula": "LC1",  "programas": "AutoCAD, MATLAB",       "estado":  "Reservada"     },
                { "id": "PC-LC2-01", "aula": "LC2",  "programas": "Python, VS Code",       "estado":  "Disponible"    },
                { "id": "PC-LC2-02", "aula": "LC2",  "programas": "Python, Java, VS Code", "estado":  "Prestada"      },
                { "id": "PC-LC2-03", "aula": "LC2",  "programas": "AutoCAD, MATLAB",       "estado":  "Disponible"    },
                { "id": "PC-LC3-01", "aula": "LC3",  "programas": "Python, VS Code",       "estado":  "Disponible"    },
                { "id": "PC-LC3-02", "aula": "LC3",  "programas": "Python, Java, VS Code", "estado":  "Prestada"      },
                { "id": "PC-LC3-03", "aula": "LC3",  "programas": "AutoCAD, MATLAB",       "estado":  "Reservada"     },
                { "id": "PC-LC3-04", "aula": "LC3",  "programas": "Python, VS Code",       "estado":  "Mantenimiento" }
              ]


prestamos = [
    { "num_control": "26040200", "programa": "python",   "aula":    "LC1", "id": "PC-LC1-01", "fecha": "12/10/2026 07:30:00"  },
    { "num_control": "26040201", "programa": "java",     "aula":    "LC1", "id": "PC-LC1-02", "fecha": "12/10/2026 08:30:00"  },
    { "num_control": "26040202", "programa": "python",   "aula":    "LC2", "id": "PC-LC2-01", "fecha": "12/10/2026 09:30:00"  },
    { "num_control": "26040203", "programa": "matlab",   "aula":    "LC3", "id": "PC-LC3-01", "fecha": "12/10/2026 10:30:00"  },
    { "num_control": "26040204", "programa": "python",   "aula":    "LC3", "id": "PC-LC3-02", "fecha": "12/10/2026 11:30:00"  }
]

necesidades = [
    { "programa": "python", "fecha": "12/10/2026 07:30:00"  },
]

# 1. Consultar computadoras
def listar_computadoras():
    print("\n--- COMPUTADORAS ---")
    for pc in computadoras:
        print("Aula:", pc["aula"], "| ID:", pc["id"], "| Software:", pc["programas"], "| Estado:", pc["estado"])


# 2. Buscar por software
def buscar_sw():
    programa = input("\nIngrese el software que desea buscar: ")
    print("\n--- RESULTADOS ---")
    encontrado = False
    for pc in computadoras:
        if programa.lower() in pc["programas"].lower():
            print("ID:", pc["id"],
                  "| Aula:", pc["aula"],
                  "| Software:", pc["programas"],
                  "| Estado:", pc["estado"])
            encontrado = True

    if not encontrado:
        print("No se encontraron computadoras con ese software.")


# 3. Solicitar préstamo
def solicitar_prestamo():
    print("\n--- SOLICITAR PRÉSTAMO ---")
    num_control = input("Ingrese el número de control: ")

    # Verificar si el alumno existe (soporta si alumnos es lista de dicts o lista simple)
    existe_alumno = False
    for alu in alumnos:
        if isinstance(alu, dict):
            if str(alu.get("num_control", alu.get("id", ""))) == num_control:
                existe_alumno = True
                break
        elif isinstance(alu, (list, tuple)):
            if str(alu[0]) == num_control:
                existe_alumno = True
                break
        else:
            if str(alu) == num_control:
                existe_alumno = True
                break

    if existe_alumno:
        programa = input("¿Qué programa necesita?: ")

        pc_encontrada = None
        # Buscar computadora que tenga el programa
        for pc in computadoras:
            if programa.lower() in pc["programas"].lower():
                pc_encontrada = pc
                if pc["estado"] == "Disponible":
                    pc["estado"] = "Prestada"
                    print(f"Se asignó la computadora {pc['id']} que está en el aula {pc['aula']}")
                    prestamos.append([num_control, programa, pc["aula"], pc["id"]])
                    return

        # Si encontró el programa pero no estaba disponible
        if pc_encontrada is not None:
            print(f"Sí se encontró el software, pero la computadora está {pc_encontrada['estado']}.")
        else:
            # Si el programa NO existe en ninguna computadora
            resp = input("El programa no existe. ¿Desea agregarlo a necesidades? (s/n): ")
            if resp.lower() == 's':
                fecha = datetime.datetime.now().strftime("%d/%m/%Y %H:%M:%S")
                necesidades.append([programa, fecha])
                print("Agregado a la lista de necesidades.")
    else:
        print("El alumno no existe.")


# 4. Devolver computadora
def devolver_computadora():
    print("\n--- DEVOLVER COMPUTADORA ---")
    id_pc = input("Dame el ID de la computadora: ")
    existe = False

    for pc in computadoras:
        if str(pc["id"]) == id_pc:
            existe = True
            if pc["estado"] in ["Prestada", "Reservada"]:
                pc["estado"] = "Disponible"
                print(f"La computadora {id_pc} ya está disponible.")
            else:
                print(f"La computadora {id_pc} no está prestada/reservada (Estado actual: {pc['estado']}).")
            break

    if not existe:
        print(f"La computadora {id_pc} no existe.")


# 6. Agregar computadora
def agregar_computadora():
    print("\n--- AGREGAR COMPUTADORA ---")
    id_computadora = input("ID de la computadora: ")
    aula = input("Aula: ")
    programas = input("Software instalado: ")

    nueva_compu = {
        "id": id_computadora,
        "aula": aula,
        "programas": programas,
        "estado": "Disponible"
    }
    computadoras.append(nueva_compu)
    print("\nComputadora agregada correctamente.")

#7
def historial_prestamos():
    print("\n--- HISTORIAL DE PRESTAMOS DE COMPUTADORAS ---")
    cantidad_prestamos = len(prestamos)
    if cantidad_prestamos > 0:
        for i in range(len(prestamos)):
            print(i+1, " | Num_control:", prestamos[i]["num_control"],
                  "| Software:", prestamos[i]["programa"],
                  " | Aula:", prestamos[i]["aula"],
                  "| ID:", prestamos[i]["id"] )
    else:
        print("No se han registrado prestamos")
#8
def historial_necesidades():
    print("\n\n=================================")
    print("Historico de necesidades")
    for i in range(len(necesidades)):
        print("Programa:", necesidades[i]["programa"], "| Fecha:", necesidades[i]["fecha"])

#9
def estadisticas():
    # Diccionario para contar préstamos por carrera
    conteo_carreras = {}
    for prestamo in prestamos:
        carrera = "Carrera no encontrada"
        for alumno in alumnos:
            if alumno["num_control"] == prestamo["num_control"]:
                carrera = alumno["carrera"]
                break
        if carrera in conteo_carreras:
            conteo_carreras[carrera] += 1
        else:
            conteo_carreras[carrera] = 1

    print("\n==============================")
    print("Estadísticas por carrera:")
    if conteo_carreras:
        for carrera, cantidad in conteo_carreras.items():
            print(carrera, ":", cantidad, "veces")
    else:
        print("No hay préstamos registrados.")

# Diccionario para contar préstamos por software solicitado
conteo_programas = {}

for prestamo in prestamos:
    programa = prestamo["programa"].lower()

    if programa in conteo_programas:
        conteo_programas[programa] += 1
    else:
        conteo_programas[programa] = 1

print("\nEstadísticas por programa:")
if conteo_programas:
    for programa, cantidad in conteo_programas.items():
        print(programa, ":", cantidad, "veces")
else:
    print("No hay préstamos registrados.")




# MENÚ PRINCIPAL

opcion = 0
while opcion <= 9:
  print("\n==========================")
  print("    SISTEMA DE PRESTAMOS    ")
  print("\n==========================")
  print("1. Consultar computadoras")
  print("2. Buscar por software")
  print("3. Solicitar prestamo")
  print("4. Devolver computadora")
  print("5. Crear reservación")
  print("6. Agregar computadora")
  print("7. Historial de Prestamos")
  print("8. Historial de Necesidades")
  print("9. Estadisticas")
  print("10. Salir")
  print("\n==========================")
    opcion = int(input("Seleccione una opción: "))


  # 1. Consultar computadora
  if opcion == 1:
     listar_computadoras()

  # 2. Buscar por software
  elif opcion == 2:
      buscar_sw()

  # 3. Solicitar préstamo
  elif opcion == 3:
    solicitar_prestamo()

  # 4. Devolver computadora
  elif opcion == 4:
      devolver_computadora()

  # 6. Agregar computadora
  elif opcion == 6:
      agregar_computadora()

  # 7. Historial de prestamos
  elif opcion == 7:
    historial_prestamos()

  # 8. Historial de necesidades
  elif opcion == 8:
    historial_necesidades()

  # 9. Estadisticas
  elif opcion == 9:
    estadisticas()

  #10. Salir
  elif opcion == 10:
    print("\nGracias por utilizar el sistema.")

    else:
    print("\nOpción no válida.")
```
